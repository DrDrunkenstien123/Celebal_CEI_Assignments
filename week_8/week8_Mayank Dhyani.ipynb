## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent

- Keyword extraction → Keyword Tool
- General queries → Direct response

````python
import re
from datetime import datetime
````

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression):
    try:
        result = eval(expression, {"__builtins__": None}, {})
        return result
    except:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text):
    stopwords = {
        "is", "the", "a", "an", "of", "and", "to", "in", "on",
        "for", "with", "from", "by", "at", "as"
    }

    words = re.findall(r'\b\w+\b', text.lower())

    keywords = [word for word in words if word not in stopwords]

    return keywords
````

````python
# 🛠️ TOOL 3: Sentiment Detector

def sentiment_tool(text):
    positive_words = ["good", "great", "happy", "excellent", "amazing", "awesome"]
    negative_words = ["bad", "sad", "terrible", "angry", "awful", "worst"]

    text = text.lower()

    pos = sum(word in text for word in positive_words)
    neg = sum(word in text for word in negative_words)

    if pos > neg:
        return "Positive"
    elif neg > pos:
        return "Negative"
    else:
        return "Neutral"
````

````python
# 🛠️ TOOL 4: Summarizer

def summarize_tool(text):
    sentences = [s.strip() for s in text.split(".") if s.strip()]

    if len(sentences) == 0:
        return "No text provided"

    if len(sentences) == 1:
        return sentences[0]

    return ". ".join(sentences[:2]) + "."
````

````python
# 🛠️ TOOL 5: Intent Detector

def detect_intent(query):
    query = query.lower()

    math_words = ["calculate", "solve", "compute", "+", "-", "*", "/"]
    keyword_words = ["keyword", "extract keywords"]
    sentiment_words = ["sentiment", "emotion"]
    summary_words = ["summarize", "summary"]

    if any(word in query for word in math_words):
        return "math"
    elif any(word in query for word in keyword_words):
        return "keyword"
    elif any(word in query for word in sentiment_words):
        return "sentiment"
    elif any(word in query for word in summary_words):
        return "summary"
    else:
        return "general"
````

````python
logs = []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)


def agent(query):
    try:
        intent = detect_intent(query)

        logs.append({
            "query": query,
            "intent": intent,
            "timestamp": str(datetime.now())
        })

        if intent == "math":
            expression = query.lower()

            for word in ["calculate", "compute", "solve"]:
                expression = expression.replace(word, "")

            expression = expression.strip()

            if not expression:
                return {
                    "status": "error",
                    "message": "No mathematical expression provided"
                }

            result = calculator(expression)

        elif intent == "keyword":
            text = re.sub(r'extract keywords from|extract keywords|keywords from',
                          '', query, flags=re.IGNORECASE).strip()

            if not text:
                return {
                    "status": "error",
                    "message": "No text provided for keyword extraction"
                }

            result = extract_keywords(text)

        elif intent == "sentiment":
            text = re.sub(r'analyze sentiment of|sentiment of|analyze sentiment',
                          '', query, flags=re.IGNORECASE).strip()

            if not text:
                return {
                    "status": "error",
                    "message": "No text provided for sentiment analysis"
                }

            result = sentiment_tool(text)

        elif intent == "summary":
            text = re.sub(r'summarize|summary of',
                          '', query, flags=re.IGNORECASE).strip()

            if not text:
                return {
                    "status": "error",
                    "message": "No text provided for summarization"
                }

            result = summarize_tool(text)

        else:
            result = f"General response: {query}"

        return {
            "query": query,
            "intent": intent,
            "result": result,
            "timestamp": str(datetime.now()),
            "status": "success"
        }

    except Exception as e:
        return {
            "status": "error",
            "message": str(e)
        }
````

## 📦 Expected Output Format

  "result": ...
}

````python
queries = [
    "Calculate 20 + 5",
    "Compute 25 * 8",
    "Solve 100 / 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "Analyze sentiment of I am happy today",
    "Analyze sentiment of Today was terrible",
    "Summarize AI is growing rapidly. It transforms industries. It improves automation.",
    "What is machine learning?",
    "Calculate",
    "Extract keywords from"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 80)
````

````python
# 🎯 Interactive Mode

while True:
    query = input("Enter query (type 'exit' to stop): ")

    if query.lower() == "exit":
        print("Assistant stopped.")
        break

    response = agent(query)
    print("Response:", response)
    print()
````
