````python
def safe_divide(a, b):
    # YOUR CODE HERE
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise TypeError("Inputs must be numeric")

    try:
        return a / b
    except ZeroDivisionError:
        return None
    #------------------------------------------------------------------------

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

**Your Geometric Explanation (4.3):**

> A matrix transformation alters the direction and magnitude of a vector. While most vectors get their direction and magnitude altered during a transformation, eigenvectors are unique vectors that preserve their direction. They simply get longer, shorter or flipped when the matrix transformation occurs.

The magnitude by which a vector is stretched or squashed is called the eigenvalue.

*   If the eigenvalue is > 1 then the vector will stretch out.
*   If the eigenvalue is between 0 and 1 then the vector will squish.
*   If the eigenvalue is negative then the direction of the vector flips.
*   If the eigenvalue is 0 then the vector has vanished ( collapsed to 0 )

Geometrically, eigenvectors are directions on which a linear transformation only scales, without producing rotation.

### 4.4 SVD & Dimensionality Reduction

1. Create a random 4×3 matrix `X` (use `np.random.seed(42)`).
2. Compute its **SVD**: `U, S, Vt = np.linalg.svd(X, full_matrices=False)`.
3. **Reconstruct** `X` from the SVD and verify it matches the original.
4. Perform a **rank-1 approximation** using only the largest singular value/vector — call it `X_approx`.
5. Explain the relationship between SVD components and PCA in the markdown cell below.

````python
np.random.seed(42)
X = np.random.randn(4, 3)

U, S, Vt = np.linalg.svd(X, full_matrices=False)  # YOUR CODE HERE

X_reconstructed = U @ np.diag(S) @ Vt  # YOUR CODE HERE — reconstruct from U, S, Vt
X_approx        = np.outer(U[:, 0] * S[0], Vt[0])  # YOUR CODE HERE — rank-1 approximation
#------------------------------------------------------------------------------------------

print(f"U shape: {U.shape}, S shape: {S.shape}, Vt shape: {Vt.shape}")
print(f"Singular values: {S}")
print(f"Reconstruction error: {np.linalg.norm(X - X_reconstructed):.2e}")
print(f"Rank-1 approximation:\n{X_approx}")

assert np.allclose(X, X_reconstructed, atol=1e-10), "Reconstruction failed!"
print("4.4 passed ✓")
````

**SVD → PCA Connection (4.4):**

> *Explain: what are U, S, Vt in SVD, and how do they relate to principal components? Which matrix contains the directions of maximum variance?*

In SVD, A matrix X can be broken down in 3 matrices:

X=U*S*V^T

Where U: Contains the left singular vectors. Think of these directions in output space (each row is a data point).

S: Contains the singular values, which tell you importance of each component. Greater singular values account for greater variance / importance.

Vt (V^T): Contains the right singular vectors. These point out directions of variation in the data (aka principal components used in PCA).

From a geometric intuition perspective V^T, tells the directions of most variation and S tells how much variation in each of those directions. U tells how to describe each point using V T, scaled by S. The matrix with directions of greatest variance is Vt (or the columns of V) as these are called the principal components (or features), of which PCA is built off of.

The rank-1 approximation is to simply take the most important feature and use it to represent your matrix.

### 5.4 Distribution Testing & Stationarity

1. Generate two samples: `s1` ~ Normal(0, 1), `s2` ~ Exponential(1), both n=200 (seed=0).
2. Run the **Kolmogorov-Smirnov test** (`stats.kstest`) comparing each against a standard normal CDF.
3. Generate a non-stationary time series (linear trend + noise). Run the **Augmented Dickey-Fuller test** (`adfuller` from `statsmodels`) and interpret the result.
4. Difference the series once and re-run ADF to confirm stationarity.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = stats.kstest(s1, 'norm')  # YOUR CODE HERE
ks_s2 = stats.kstest(s2, 'norm')  # YOUR CODE HERE
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = adfuller(ts)  # YOUR CODE HERE — adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff = np.diff(ts)  # YOUR CODE HERE — np.diff
adf_diff = adfuller(ts_diff)  # YOUR CODE HERE
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

**Interpretation (5.4)**

KS Test for s1 (Normal Distribution):
The p-value is 0.6268, which is greater than 0.05. Therefore, we fail to reject the null hypothesis, meaning the sample is consistent with a normal distribution.

KS Test for s2 (Exponential Distribution):
The p-value is 0.0000, which is less than 0.05. Therefore, we reject the null hypothesis, meaning the sample does not follow a normal distribution.

ADF Test on Original Time Series:
The p-value is 0.7802, which is greater than 0.05. Hence, the series is non-stationary because it contains a trend over time.

ADF Test After Differencing:
After applying first-order differencing, the p-value becomes 0.0000, which is less than 0.05. Therefore, the differenced series is stationary. This shows that differencing successfully removed the trend and stabilized the series.

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    # YOUR CODE HERE
    epsilon = 1e-10

    # Create bins using expected distribution percentiles
    breakpoints = np.linspace(0, 100, bins + 1)
    breakpoints = np.percentile(expected, breakpoints)

    # Count values in each bin
    expected_counts, _ = np.histogram(expected, bins=breakpoints)
    actual_counts, _ = np.histogram(actual, bins=breakpoints)

    # Convert counts to percentages
    expected_pct = expected_counts / len(expected)
    actual_pct = actual_counts / len(actual)

    # Compute PSI
    psi = np.sum((actual_pct - expected_pct) *np.log((actual_pct + epsilon) / (expected_pct + epsilon)))

    return psi

np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions
fig, ax = plt.subplots()
# YOUR CODE HERE
ax.hist(train_dist, bins=30, alpha=0.6, density=True, label='Training Distribution')
ax.hist(drift_dist, bins=30, alpha=0.6, density=True, label='Production Distribution')
#-------------------------------------------------------------------------------------------
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

- **Concept drift:** The relationship between input features and the target variable changes over time, reducing model accuracy.
- **Covariate drift:** The distribution of input features changes over time while the relationship with the target may remain unchanged.
- **PSI < 0.1:** No significant shift, the model input distribution is stable.
- **PSI 0.1–0.2:** Minor shift, monitor the model closely for possible degradation.
- **PSI > 0.2:** Major shift, the model may no longer perform reliably and retraining is recommended.
- **Retraining trigger example:** Retrain the model when PSI exceeds 0.2 or when production accuracy drops significantly compared to training performance.

   - P(first=red, second=blue) — **joint probability**
   - P(second=blue | first=red) — **conditional probability**
3. Are the two draws **independent**? Show mathematically.

````python
total = 10
red, blue, green = 4, 3, 3

p_red   = red / total    # YOUR CODE HERE
p_blue  = blue / total   # YOUR CODE HERE
p_green = green / total  # YOUR CODE HERE

# Joint: P(red first AND blue second)
p_red_then_blue = (red / total) * (blue / (total - 1))  # YOUR CODE HERE

# Conditional: P(blue second | red first)
p_blue_given_red = blue / (total - 1)  # YOUR CODE HERE

# Independence check: are P(B|A) == P(B) ?
independent = p_blue_given_red == p_blue  # YOUR CODE HERE — True/False
#----------------------------------------------------------------------------

print(f"P(red)={p_red:.3f}, P(blue)={p_blue:.3f}, P(green)={p_green:.3f}")
print(f"P(red,blue): {p_red_then_blue:.4f}")
print(f"P(blue|red): {p_blue_given_red:.4f}")
print(f"P(blue):     {p_blue:.4f}")
print(f"Independent? {independent}")
````

**Mathematical Explanation**

P(red)= 4/10

P(blue)= 3/10

P(green)= 3/10

Joint probability: P(red first AND blue second)= (4/10)*(3/9)= 12/90 = 0.1333

Conditional probability: P(blue second∣red first)= 3/9 = 0.3333

Independence check: P(blue)= 3/10 = 0.3

Since, P(blue second|red first)!=P(blue) the two draws are not independent because drawing without replacement changes the probabilities.

### 6.3 Bayes' Theorem

**Scenario — Spam filter:**
- P(Spam) = 0.30 (prior)
- P("free" | Spam) = 0.80 (likelihood)
- P("free" | Not Spam) = 0.05

1. Compute P(Spam | "free") using Bayes' theorem.
2. Implement a simple `naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham)` that returns the posterior P(Spam | word).
3. In the markdown cell, map each term — **prior, likelihood, evidence, posterior** — to the spam example.

````python
p_spam            = 0.30
p_free_given_spam = 0.80
p_free_given_ham  = 0.05
p_ham             = 1 - p_spam

# P(free) — law of total probability
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)  # YOUR CODE HERE

# Bayes
p_spam_given_free = (p_free_given_spam * p_spam) / p_free  # YOUR CODE HERE

print(f"P('free'): {p_free:.4f}")
print(f"P(Spam | 'free'): {p_spam_given_free:.4f}")

def naive_bayes_predict(prior_spam, p_word_given_spam, p_word_given_ham):
    # YOUR CODE HERE
    p_ham = 1 - prior_spam
    p_word = (p_word_given_spam * prior_spam) + (p_word_given_ham * p_ham)
    posterior = (p_word_given_spam * prior_spam) / p_word

    return posterior
#---------------------------------------------------------------------------------------------
pred = naive_bayes_predict(0.30, 0.80, 0.05)
assert abs(pred - p_spam_given_free) < 1e-9
print(f"naive_bayes_predict: {pred:.4f}")
print("6.3 passed ✓")
````

**Bayes Term Mapping (6.3):**

| Term | Formula | In this example |
|------|---------|-----------------|
| Prior | P(Spam) | Probability that an email is spam before seeing the word “free” |
| Likelihood | P(word \| Spam) | Probability that the word “free” appears in a spam email |
| Evidence | P(word) | Overall probability of seeing the word “free” in any email |
| Posterior | P(Spam \| word) | Probability that the email is spam after seeing the word “free” |

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30
sample_means = np.array([np.mean(np.random.choice(population, sample_size))for i in range(n_samples)])  # YOUR CODE HERE — array of 5000 sample means

pop_mean = population.mean()
pop_std  = population.std()
clt_std  = pop_std / np.sqrt(sample_size)  # YOUR CODE HERE — predicted std of sample means

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
# YOUR CODE HERE — histogram of sample means + theoretical normal overlay
# Histogram of sample means
ax.hist(sample_means, bins=30, density=True, alpha=0.6, label='Sample Means')

# Theoretical normal distribution
x = np.linspace(sample_means.min(), sample_means.max(), 300)

ax.plot(x,stats.norm.pdf(x, pop_mean, clt_std),label='CLT Normal Distribution')

ax.legend()
#----------------------------------------------------------------------------------------

ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

**CLT Reflection (6.4):**

> *In 2–3 sentences: Why does this result matter for statistics and machine learning? What assumption does the CLT justify in many classical statistical tests?*

The Central Limit Theorem is important because it shows that the distribution of sample means becomes approximately normal as the sample size increases, even if the original population is not normally distributed. This allows statisticians and machine learning practitioners to use normal-distribution-based methods for estimation, confidence intervals, and hypothesis testing.

The CLT justifies the assumption of normality used in many classical statistical tests such as t-tests, z-tests, and linear regression analysis.
